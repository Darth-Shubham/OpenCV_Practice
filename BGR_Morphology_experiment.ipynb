```python
import cv2
import numpy as np
```

```python
img_ori = cv2.imread('hist2.jpg')
b, g, r = cv2.split(img_ori)
kernel = np.ones((5,5))
```

```python
def img_operation(b, g, r, b_morph, g_mprph, r_morph):
    img_b_morph = cv2.merge((b_morph,g,r))
    img_g_morph = cv2.merge((b,g_mprph,r))
    img_r_morph = cv2.merge((b,g,r_morph))
    img_bgr_morph = cv2.merge((b_morph, g_mprph, r_morph))
    
    return img_b_morph, img_g_morph, img_r_morph, img_bgr_morph
```

```python
b_dilate = cv2.morphologyEx(b, cv2.MORPH_DILATE, kernel, iterations = 3)
g_dilate = cv2.morphologyEx(g, cv2.MORPH_DILATE, kernel, iterations = 3)
r_dilate = cv2.morphologyEx(r, cv2.MORPH_DILATE, kernel, iterations = 3)

b_erode = cv2.morphologyEx(b, cv2.MORPH_ERODE, kernel, iterations = 1)
g_erode = cv2.morphologyEx(g, cv2.MORPH_ERODE, kernel, iterations = 1)
r_erode = cv2.morphologyEx(r, cv2.MORPH_ERODE, kernel, iterations = 1)

b_open = cv2.morphologyEx(b, cv2.MORPH_OPEN, kernel, iterations = 1)
g_open = cv2.morphologyEx(g, cv2.MORPH_OPEN, kernel, iterations = 1)
r_open = cv2.morphologyEx(r, cv2.MORPH_OPEN, kernel, iterations = 1)

b_close = cv2.morphologyEx(b, cv2.MORPH_CLOSE, kernel, iterations = 1)
g_close = cv2.morphologyEx(g, cv2.MORPH_CLOSE, kernel, iterations = 1)
r_close = cv2.morphologyEx(r, cv2.MORPH_CLOSE, kernel, iterations = 1)

b_grad = cv2.morphologyEx(b, cv2.MORPH_GRADIENT, kernel, iterations = 1)
g_grad = cv2.morphologyEx(g, cv2.MORPH_GRADIENT, kernel, iterations = 1)
r_grad = cv2.morphologyEx(r, cv2.MORPH_GRADIENT, kernel, iterations = 1)
```

```python
new_b, new_g, new_r, new_bgr = img_operation(b,g,r,b_dilate,g_dilate,r_dilate)
```

```python
cv2.imshow('ori', img_ori)
cv2.imshow('change B', new_b)
cv2.imshow('change G', new_g)
cv2.imshow('change R', new_r)
cv2.imshow('change BGR', new_bgr)

if cv2.waitKey(0) & 0xff == ord('q'):
    cv2.destroyAllWindows()
```
