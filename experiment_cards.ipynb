```python
import cv2
import numpy as np
```

```python
image1 = cv2.imread("cards1.jpg")
image2 = cv2.imread("cards2.jpg")
image3 = cv2.imread("cards3.jpg")
```

```python
image1_grey = cv2.cvtColor(image1, cv2.COLOR_BGR2GRAY)
image2_grey = cv2.cvtColor(image2, cv2.COLOR_BGR2GRAY)
image3_grey = cv2.cvtColor(image3, cv2.COLOR_BGR2GRAY)
```

```python
lthresh_image1, hthresh_image1 = 50,50
lthresh_image2, hthresh_image2 = 50,50
lthresh_image3, hthresh_image3 = 50,50
```

```python
def lowerTrhesh1(lowerThreshold1):
    global lthresh_image1
    lthresh_image1 = lowerThreshold1

def lowerTrhesh2(lowerThreshold2):
    global lthresh_image2
    lthresh_image2 = lowerThreshold2

def lowerTrhesh3(lowerThreshold3):
    global lthresh_image3
    lthresh_image3 = lowerThreshold3
```

```python
def higherTrhesh1(higherThreshold1):
    global hthresh_image1
    hthresh_image1 = higherThreshold1

def higherTrhesh2(higherThreshold2):
    global hthresh_image2
    hthresh_image2 = higherThreshold2

def higherTrhesh3(higherThreshold3):
    global hthresh_image3
    hthresh_image3 = higherThreshold3
```

```python
cv2.namedWindow("TrackBars", cv2.WINDOW_FREERATIO)
cv2.createTrackbar("lThresh1", "TrackBars", 34,200, lowerTrhesh1)
cv2.createTrackbar("hThresh1", "TrackBars", 119,200, higherTrhesh1)
cv2.createTrackbar("lThresh2", "TrackBars", 89,200, lowerTrhesh2)
cv2.createTrackbar("hThresh2", "TrackBars", 200,200, higherTrhesh2)
cv2.createTrackbar("lThresh3", "TrackBars", 124,200, lowerTrhesh3)
cv2.createTrackbar("hThresh3", "TrackBars", 200,200, higherTrhesh3)


while True:

    canny1 = cv2.Canny(image1_grey, lthresh_image1, hthresh_image1)
    canny2 = cv2.Canny(image2_grey, lthresh_image2, hthresh_image2)
    canny3 = cv2.Canny(image3_grey, lthresh_image3, hthresh_image3)
    
    cv2.imshow("Card 1", canny1)
    cv2.imshow("Card 2", canny2)
    cv2.imshow("Card 3", canny3)

    t_lthresh_img1 = cv2.getTrackbarPos("lThresh1", "TrackBars")
    t_hthresh_img1 = cv2.getTrackbarPos("hThresh1", "TrackBars")
    t_lthresh_img2 = cv2.getTrackbarPos("lThresh2", "TrackBars")
    t_hthresh_img2 = cv2.getTrackbarPos("hThresh2", "TrackBars")
    t_lthresh_img3 = cv2.getTrackbarPos("lThresh3", "TrackBars")
    t_hthresh_img3 = cv2.getTrackbarPos("hThresh3", "TrackBars")
    
    if cv2.waitKey(1) & 0xff == ord('q'):
        break
cv2.destroyAllWindows()
```

```python
kernel = np.ones((3,3))

clean_edge_can1 = cv2.morphologyEx(canny1, cv2.MORPH_CLOSE,kernel)
clean_edge_can2 = cv2.morphologyEx(canny2, cv2.MORPH_CLOSE,kernel)
clean_edge_can3 = cv2.morphologyEx(canny3, cv2.MORPH_CLOSE,kernel)


cv2.imshow("Card 1", clean_edge_can1)
cv2.imshow("Card 2", clean_edge_can2)
cv2.imshow("Card 3", clean_edge_can3)

if cv2.waitKey(0) & 0xff == ord('q'):
    cv2.destroyAllWindows()
```

```python
contour1, _ = cv2.findContours(canny1, mode = cv2.RETR_EXTERNAL, method = cv2.CHAIN_APPROX_SIMPLE)
contour2, _ = cv2.findContours(canny2, mode = cv2.RETR_EXTERNAL, method = cv2.CHAIN_APPROX_SIMPLE) 
contour3, _ = cv2.findContours(canny3, mode = cv2.RETR_EXTERNAL, method = cv2.CHAIN_APPROX_SIMPLE) 

contours1 = cv2.drawContours(canny1.copy(), contour1, -1, (255), thickness = 2, lineType = cv2.LINE_AA)
contours2 = cv2.drawContours(canny2.copy(), contour2, -1, (255), thickness = 2, lineType = cv2.LINE_AA)
contours3 = cv2.drawContours(canny3.copy(), contour3, -1, (255), thickness = 2, lineType = cv2.LINE_AA)


cv2.imshow("Card Contour 1", contours1)
cv2.imshow("Card Contour 2", contours2)
cv2.imshow("Card Contour 3", contours3)
if cv2.waitKey(0) & 0xff == ord('q'):
    cv2.destroyAllWindows()
```

```python
max_cont1 = max(contour1, key = cv2.contourArea)
max_cont2 = max(contour2, key = cv2.contourArea)
max_cont3 = max(contour3, key = cv2.contourArea)


# Finding the coordinated of the bounding box
x1,y1,w1,h1 = cv2.boundingRect(max_cont1)
x2,y2,w2,h2 = cv2.boundingRect(max_cont2)
x3,y3,w3,h3 = cv2.boundingRect(max_cont3)
```

```python
img_bound1 = cv2.rectangle(image1, (x1,y1), (x1+w1, y1+h1), (0, 255 , 0), 4)
img_bound2 = cv2.rectangle(image2, (x2,y2), (x2+w2, y2+h2), (0, 255 , 0), 4)
img_bound3 = cv2.rectangle(image3, (x3,y3), (x3+w3, y3+h3), (0, 255 , 0), 4)

cv2.imshow("Bounded Card 1", img_bound1)
cv2.imshow("Bounded Card 2", img_bound2)
cv2.imshow("Bounded Card 3", img_bound3)
if cv2.waitKey(0) & 0xff == ord('q'):
    cv2.destroyAllWindows()
```
