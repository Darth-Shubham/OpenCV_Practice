```python
import cv2
import numpy as np
```

# Phase 1

```python
# Reading image
ori_img = cv2.imread('face2.jpg')
ori_img = cv2.resize(ori_img, None, fx=.75, fy = .65)
```

# Phase 2

```python
# Removing Impurities using median filter
image_cleaned = cv2.medianBlur(ori_img, 7)
cv2.imshow("median blur 1", ori_img)
image_cleaned = cv2.medianBlur(image_cleaned, 9)
# cv2.imshow("median blur 2", image_cleaned)
image_cleaned = cv2.medianBlur(image_cleaned, 15)
cv2.imshow("median blur 3", image_cleaned)

# Preserving the edges in the filtered image
iamge_cleared = cv2.edgePreservingFilter(image_cleaned, sigma_s = 15)
cv2.imshow("Cleaned image", image_cleaned)
```

# Phase 3

```python
# Using Bilateral filtering
image_filter = cv2.bilateralFilter(image_cleaned, 5, 5, 5)
img_flag = 0
for i in range(2):
    image_filter = cv2.bilateralFilter(image_filter, 9, 7, 7)
        
cv2.imshow("Bilateral Filter {}".format(img_flag), image_filter)
img_flag +=1

for i in range(3):
    image_filter = cv2.bilateralFilter(image_filter, 15, 11, 11)
    
cv2.imshow("Bilateral Filter {}".format(img_flag), image_filter)
img_flag +=1
```

# Phase 4

```python
# Sharpening the image
gaussian_mask = cv2.GaussianBlur(image_filter, (9,9), 13)
image_sharp = cv2.addWeighted(image_filter, 1.89, gaussian_mask, -.89, 0)
cv2.imshow("Image Sharp", image_sharp)
image_sharp = cv2.addWeighted(image_filter, 1.56, gaussian_mask, -.56, 5)
```

```python
cv2.imshow("Original Image", ori_img)
cv2.imshow("Oil Painting", image_sharp)
```
