```python
import cv2
import numpy as np
```

```python
draw = False
mode = False
erase = False
a, b = -1,-1
```

```python
def onChange(value):
    global mode
    if value == 0:
        mode = False
    elif value == 1:
        mode = True

def eraser(valueq):
    global erase
    if valueq == 0:
        erase = False
    elif valueq == 1:
        erase = True
    
def drawCircle(event, x, y, flags, params):
    global draw,mode, a,b, img, erase
    
    red = cv2.getTrackbarPos('Red', 'Trackbar')
    green = cv2.getTrackbarPos('Green', 'Trackbar')
    blue = cv2.getTrackbarPos('Blue', 'Trackbar')
    
    if event == cv2.EVENT_LBUTTONDOWN:
        draw = True
        a, b = x, y
        
    elif event == cv2.EVENT_MOUSEMOVE:
        if draw:
            if mode:
                # cv2.circle(img, (x, y), 2, (0, 255, 0), -1)  # Draw line using circle
                cv2.circle(img, (a, b), abs(x - a), (blue,green,red), 2)  # Draw circle
            else:
                cv2.rectangle(img, (a, b), (x, y), (blue,green,red), -1)  # Draw rectangle
            
    elif event == cv2.EVENT_LBUTTONUP:
        draw = False
        if mode:
            # cv2.circle(img, (x, y), 2, (0, 255, 0), -1)  # Draw line using circle
            cv2.circle(img, (a, b), abs(x - a), (blue,green,red), -1)  # Final circle
        else:
            cv2.rectangle(img, (a, b), (x, y), (blue,green,red), -1)  # Final rectangle

    elif event == cv2.EVENT_LBUTTONDBLCLK:
        if erase:
            img[:] = 255
```

```python
# Creating the canvas Matrix
img = np.ones((512, 512, 3), np.uint8) * 255

# Macking canvas background white
# img[:,:] = img[:,:]+255

# Creating Trackbar window
cv2.namedWindow("Trackbar")
#Creating Trackbar
cv2.createTrackbar('Shape','Trackbar', 0, 1, onChange)
cv2.createTrackbar('Erase','Trackbar', 0, 1, eraser)

cv2.createTrackbar('Red','Trackbar', 0, 255, TrackPass)
cv2.createTrackbar('Green','Trackbar', 0, 255, TrackPass)
cv2.createTrackbar('Blue','Trackbar', 0, 255, TrackPass)


# Creating Canvas window
cv2.namedWindow("Paint")
# Making Event listner
cv2.setMouseCallback('Paint', drawCircle)

# Showing Canvas
while True:
    cv2.imshow('Paint', img)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cv2.destroyAllWindows()
```
