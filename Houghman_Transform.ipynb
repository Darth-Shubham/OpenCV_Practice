# Standard Hough Line Method

```python
# Detecting Lines
houghLines = cv2.HoughLines(cannyEdge, 1, numpy.pi/180, 150)
```

```python
for line in houghLines:
    rho, theta = line[0]
    a = numpy.cos(theta)
    b = numpy.sin(theta)
    x0 = a*rho
    y0 = b*rho

    x1 = int(x0 + 800 * (-b))
    y1 = int(y0 + 800 * (a))

    x2 = int(x0 - 800 * (-b))
    y2 = int(y0 - 800 * (a))

    cv2.line(img, (x1, y1), (x2, y2), (0,255,0), 1, cv2.LINE_AA)
```

# Houghman Project

```python
def region_of_intrest(img, roi):
    roi = [roi]
    colour_channel = img.shape[2]
    mask = numpy.zeros_like(img)
    match_mask_col = (255,) * colour_channel
    cv2.fillPoly(mask, roi, match_mask_col)
    masked_img = cv2.bitwise_and(img, mask)

    return masked_img
```

```python
def draw_lines(img ,lines):
    img = numpy.copy(img)
    blank_img = numpy.zeros((img.shape[0], img.shape[1], 3), dtype = numpy.uint8)

    for line in lines:
        for x1,y1,x2,y2 in line:
            cv2.line(blank_img, (x1,y1), (x2,y2), (0,255,0), 2)
    img = cv2.addWeighted(img, .8, blank_img, 1, 0)
    return img
```
