```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
# Reading the image
img = cv2.imread('hist1.jpg')
# Making image Grey-scale
img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
histogram = cv2.calcHist(img, [0], None, [256], [0,255])
plt.plot(histogram)
```

# Global Thresholdings

## Manual Thresholdings

```python
thresh_val, thresh_Binary_100 = cv2.threshold(img, 100, 255, cv2.THRESH_BINARY)
thresh_val, thresh_Binary_135 = cv2.threshold(img, 135, 255, cv2.THRESH_BINARY)
thresh_val, thresh_Binary_Inv = cv2.threshold(img, 135, 255, cv2.THRESH_BINARY_INV)
thresh_val, thresh_Trunc = cv2.threshold(img, 135, 255, cv2.THRESH_TRUNC)
thresh_val, thresh_toZero = cv2.threshold(img, 135, 255, cv2.THRESH_TOZERO)
thresh_val, thresh_toZero_Inv = cv2.threshold(img, 135, 255, cv2.THRESH_TOZERO_INV)
```

## Otsu Thresholding

```python
otsu_threshold, otsu_thresh_img = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
```

## Triangle Thresholding

Applied in Images which has Multiple peaks in histogram

```python
triangle_threshold, triangle_thresh_img = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_TRIANGLE)
```

# Adaptive Thresholdings

## Mean & Gaussian Thresholding

```python
adap_mean_c = cv2.adaptiveThreshold(img, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 25, 2)
adap_gaus_c = cv2.adaptiveThreshold(img, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 25, 3)
```

## 

```python
cv2.imshow('a', img)
# Global Thresholding
# Manual
cv2.imshow('Threshold 100', thresh_Binary_100)
cv2.imshow('Threshold 135', thresh_Binary_135)
cv2.imshow('Threshold Binary Inv', thresh_Binary_Inv)
cv2.imshow('Threshold Trunc', thresh_Trunc)
cv2.imshow('Threshold ToZero', thresh_toZero)
cv2.imshow('Threshold ToZero Inv', thresh_toZero_Inv)
# Otsu
cv2.imshow("Otsu Thresholding", otsu_thresh_img)
# Triangle
cv2.imshow("Triangle Thresholding", triangle_thresh_img)

# Adaptive Thresholding
cv2.imshow("Adaptive Mean C", adap_mean_c)
cv2.imshow("Adaptive Gaussian C", adap_gaus_c)


if cv2.waitKey(0) & 0xFF == ord('q'):
    cv2.destroyAllWindows()
```
