```python
import cv2
import numpy as np
```

```python
img = cv2.imread('face2.jpg')
img = cv2.resize(img, None, fx = .55, fy = .55)
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
lThreshold, hThreshold = 50,50
```

```python
def Lower_Threshold(lower_val):
    global lThreshold
    lThreshold = lower_val

def Higher_Threshold(higher_val):
    global hThreshold
    hThreshold = higher_val
```

```python
# Perfoeming Edeg Detection using Sobel X & Sobel Y
sobelX = cv2.Sobel(img_gray, -1, 1, 0)
sobelY = cv2.Sobel(img_gray, -1, 0, 1)

# Adding 50% Sobel X and 50% Sobel Y using AddWighted function
sobelXY = cv2.addWeighted(sobelX, .5, sobelY, .5, 0)

# Using Laplacian Edge detection to find gradient
laplacian = cv2.Laplacian(img_gray, -1)
```

```python
# Creating Threshold Trackbars
cv2.namedWindow("Threshold")
cv2.createTrackbar('Lower_Threshold', 'Threshold', 50, 200, Lower_Threshold)
cv2.createTrackbar('Higher_Threshold', 'Threshold', 50, 200, Higher_Threshold)

while True:

    # Using Canny Edge to enhance the relevent edges and removing unnecessary edges using hysterisis
    # Using trackbars to test various thresholding limits
    cannyEdge = cv2.Canny(img_gray, lThreshold, hThreshold)
    cv2.imshow('Original', img)
    cv2.imshow('Soble-X', sobelX)
    cv2.imshow('Soble-Y', sobelY)
    cv2.imshow('Soble-X+Y', sobelXY)
    cv2.imshow('Gradient', laplacian)
    cv2.imshow('Canny Edge Detection', cannyEdge)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cv2.destroyAllWindows()
```
