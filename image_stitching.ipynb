```python
import cv2
import numpy as np
import glob
import imutils
```

```python
image_path = glob.glob("images/capture*.jpg")
images = []
```

```python
for image in image_path:
    img = cv2.imread(image)
    # img = cv2.resize(img, None, fx = .2, fy = .2)
    images.append(img)
    cv2.imshow("image",img)
    cv2.waitKey(0)
```

```python
# Creating stitcher object
imsticher = cv2.Stitcher_create()
```

```python
error, stichedImg = imsticher.stitch(images)
```
