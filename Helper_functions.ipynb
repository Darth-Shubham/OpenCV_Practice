```python
import cv2
import datetime
import math
%run Constants.ipynb
COLOURS
RADIUS
CENTER
CANVAS
```

```python
def get_ticks():
    hours_init = []
    hours_dest = []

    for i in range(0,360,6):
        x_coord = int(CENTER[0] + (RADIUS-10) * math.cos(i*math.pi/180))
        y_coord = int(CENTER[1] + (RADIUS-10) * math.sin(i*math.pi/180))

        hours_init.append((x_coord,y_coord))

    for i in range(0,360,6):
        x_coord = int(CENTER[0] + (RADIUS-30) * math.cos(i*math.pi/180))
        y_coord = int(CENTER[1] + (RADIUS-30) * math.sin(i*math.pi/180))

        hours_dest.append((x_coord,y_coord))

    return hours_init, hours_dest        
```

```python
def getDigitalTime(h,m,s):
    time = ""
    hours = ""
    minutes = ""
    seconds = ""
    if h<10:
        hours = "0{}".format(h)
    else:
        hours = "{}".format(h)
    if m<10:
        minutes = "0{}".format(m)
    else:
        minutes = "{}".format(m)
    if s<10:
        seconds = "0{}".format(s)
    else:
        seconds = "{}".format(s)
    time = "{}:{}:{}".format(hours, minutes, seconds)
    return time
```

```python
def draw_time(image):
    time_now = datetime.datetime.now().time()
    hours = math.fmod(time_now.hour, 12)
    minute = time_now.minute
    second = time_now.second

    second_angle = math.fmod(second * 6 + 270, 360)
    minute_angle = math.fmod(minute * 6 + 270, 360)
    hour_angle = math.fmod((hours*30) + (minute/2) + 270, 360)

    sec_x = int(CENTER[0] + (RADIUS-50) * math.cos(second_angle*math.pi/180))
    sec_y = int(CENTER[1] + (RADIUS-50) * math.sin(second_angle*math.pi/180))
    cv2.line(image, CENTER, (sec_x, sec_y), COLOURS['amber'] ,1)

    min_x = int(CENTER[0] + (RADIUS-50) * math.cos(minute_angle*math.pi/180))
    min_y = int(CENTER[1] + (RADIUS-50) * math.sin(minute_angle*math.pi/180))
    cv2.line(image, CENTER, (min_x, min_y), COLOURS['black'] ,2)

    hour_x = int(CENTER[0] + (RADIUS-100) * math.cos(hour_angle*math.pi/180))
    hour_y = int(CENTER[1] + (RADIUS-100) * math.sin(hour_angle*math.pi/180))
    cv2.line(image, CENTER, (hour_x, hour_y), COLOURS['black'] ,2)

    cv2.circle(image, CENTER, 5, COLOURS['black'], -1)
    cv2.circle(image, CENTER, 3, COLOURS['white'], -1)

    time = getDigitalTime(int(hours), minute, second)
    cv2.putText(image, time, (180, 320), cv2.FONT_HERSHEY_COMPLEX, 1.1, COLOURS['red'], 1, cv2.LINE_AA)

    return image
```
