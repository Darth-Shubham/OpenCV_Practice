```python
import cv2
import numpy as np
```

```python
def setVal(val):
    print("")
```

```python
def trackbars():
    cv2.namedWindow("Trackbars")
    cv2.createTrackbar("Upper Hue", "Trackbars", 53, 180, setVal)
    cv2.createTrackbar("Upper Saturation", "Trackbars", 255, 255, setVal)
    cv2.createTrackbar("Upper Value", "Trackbars", 182, 255, setVal)
    cv2.createTrackbar("Lower Hue", "Trackbars", 12, 180, setVal)
    cv2.createTrackbar("Lower Saturation", "Trackbars", 17, 255, setVal)
    cv2.createTrackbar("Lower Value", "Trackbars", 21, 255, setVal)
```

```python
def get_frame_info(cap, scale_factor):
    success, img = cap.read()
    img = cv2.resize(img, None, fx = scale_factor, fy = scale_factor, interpolation = cv2.INTER_AREA)
    return img
```

```python
def maskOperation(mask):
    kernel = np.ones((7,7), np.uint8)
    opening = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)
    closing = cv2.morphologyEx(opening, cv2.MORPH_CLOSE, kernel)
    return opening, closing, erosion, dilation
```

```python
if __name__ == '__main__':
    cap = cv2.VideoCapture(0)
    scaleing_factor = 1.3
    trackbars()
    init_frame = get_frame_info(cap, scaleing_factor)
    while True:
        img = get_frame_info(cap, scaleing_factor)
        hsv_img = cv2.cvtColor(init_frame, cv2.COLOR_BGR2HSV)

        u_hue = cv2.getTrackbarPos("Upper Hue", "Trackbars")
        u_saturation = cv2.getTrackbarPos("Upper Saturation", "Trackbars")
        u_value = cv2.getTrackbarPos("Upper Value", "Trackbars")

        l_hue = cv2.getTrackbarPos("Lower Hue", "Trackbars")
        l_saturation = cv2.getTrackbarPos("Lower Saturation", "Trackbars")
        l_value = cv2.getTrackbarPos("Lower Value", "Trackbars")

        upper_hsv = np.array([[u_hue, u_saturation, u_value]])
        lower_hsv = np.array([[l_hue, l_saturation, l_value]])

        mask = cv2.inRange(hsv_img, lower_hsv, upper_hsv)

        bgr_img = cv2.bitwise_and(init_frame, init_frame, mask = mask)
        bgr_img = cv2.medianBlur(bgr_img, 5)
        
        cv2.imshow("original", img)
        cv2.imshow("mask", mask)
        cv2.imshow('Frame', init_frame)
        cv2.imshow("image", bgr_img)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

cap.release()
cv2.destroyAllWindows()
```
