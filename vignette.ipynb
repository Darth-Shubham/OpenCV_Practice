```python
import cv2
import numpy as np
```

```python
img = cv2.imread('face2.jpg')
img = cv2.resize(img, None, fx = .50 ,fy=.50)
height, width = img.shape[:2]
```

# Creating Vignette

```python
# Generating vignette mask using gausian kernel
kernel_x = cv2.getGaussianKernel(width, 180)
kernel_y = cv2.getGaussianKernel(height, 150)
kernel = kernel_y*kernel_x.T

# Normalizing Kernal
kernel = kernel/np.linalg.norm(kernel)

# Gererating image Mask
mask = kernel*255
final_img = np.copy(img)

for i in range(3):
    final_img[:,:,i] = final_img[:,:,i] * mask
```

```python
cv2.imshow('image', img)
cv2.imshow('vignet', final_img)
cv2.imshow('masked image', mask)


if cv2.waitKey(0) & 0xFF == ord('q'):
    cv2.destroyAllWindows()
```

# Vignette Focus and Radius using Trackbars

```python
focus_x = width//2
focus_y = height//2
radius = 190
value = 5
```

```python
def changeFocusX(value_x):
    global focus_x
    focus_x = value_x

def changeFocusY(value_y):
    global focus_y
    focus_y = value_y

def chageRadius(value_r):
    global radius
    radius = value_r

def changeVal(value_val):
    global value
    value = value_val
```

```python
cv2.namedWindow("TrackBar")
cv2.createTrackbar("Width", 'TrackBar', int(width//2), width, changeFocusX)
cv2.createTrackbar("Height", 'TrackBar', int(height//2), height, changeFocusY)
cv2.createTrackbar("Radius", 'TrackBar', radius, width, chageRadius)
cv2.createTrackbar("value", 'TrackBar', 1, 10, changeVal)


while True:
    kernel_x = cv2.getGaussianKernel(int(width), radius)
    kernel_y = cv2.getGaussianKernel(int(height), radius)
    kernel = kernel_y * kernel_x.T
    kernel = kernel/np.linalg.norm(kernel)

   
    
    shifted_kernel = np.roll(kernel, (focus_y - height // 2, focus_x - width // 2), axis=(0, 1))    
    # shifted_kernel = mask[
    #     max(int(0.1 * value * height), focus_y - height // 2):min(int(0.1 * value * height) + height, focus_y + height // 2),
    #     max(int(0.1 * value * width), focus_x - width // 2):min(int(0.1 * value * width) + width, focus_x + width // 2)
    # ]
    mask = shifted_kernel * 255
    final_img = np.copy(img)
    for i in range(3):
        final_img[:,:,i] = final_img[:,:,i] * mask

    cv2.imshow('image', img)
    cv2.imshow('masked image', final_img)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cv2.destroyAllWindows()
    
```
