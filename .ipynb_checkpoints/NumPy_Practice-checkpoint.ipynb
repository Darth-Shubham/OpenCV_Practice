# Indexing in Numpy

```python
# 4D array
arr = np.array([
    [[[1,2,3,4], [4,5,6,7],[7,8,9,0]]], 
    [[[11,12,13,14], [14,15,16,17],[17,18,19,10]]], 
    [ [ [21,22,23,24], [24,25,26,27],[27,28,29,20] ] ]
] )
print(arr)
```

```python
 #1stD,2ndD,3rdD,4thD
#  2    0    2   1:3
# 1st Dimension has 3 elements
'''[

   [[[ 1  2  3  4]
   [ 4  5  6  7]
   [ 7  8  9  0]]]


 [[[11 12 13 14]
   [14 15 16 17]
   [17 18 19 10]]]


 [[[21 22 23 24]
   [24 25 26 27]
   [27 28 29 20]]]
   
]'''
# 2st Dimension has 1 elements each
'''[[
   [[ 1  2  3  4]
   [ 4  5  6  7]
   [ 7  8  9  0]]
   ]


 [
 [[11 12 13 14]
   [14 15 16 17]
   [17 18 19 10]]
 ]


 [
 [[21 22 23 24]
   [24 25 26 27]
   [27 28 29 20]]
]
]'''

# 3rd Dimension has 3 elements each
'''
   [ 1  2  3  4]
   [ 4  5  6  7]
   [ 7  8  9  0]
'''
# 4th Dimension has 4 elements each
'''
   [ 1  2  3  4]
'''

# Slicing
arr[2 , 0, 2, 1:3]
```

```python
arr.shape
```

# Reshape

```python
re_arr = arr
re_arr = re_arr.reshape(6, 6)
print(re_arr)
```

# Flatern Matrix

```python
# Creates a 1D vector of the matrix
fl_arr = re_arr
fl_arr = fl_arr.ravel()
print(fl_arr)
```

# Creating a grey image

```python
image = np.zeros((250,510))
image.shape
```

```python
image[:,:] = 100
# image = image[:,:] + 10
# OR
# image[:,:] =110

# Creating White rectangels in the grey image
# height = image.max()
# image[int(image.shape[0]/2) - 25 :int(image.shape[0]/2) + 26, int(image.shape[1]/2) - 25 :int(image.shape[1]/2)+26] = 255
```

```python
cv2.imwrite('sample.jpg', image)
```

```python
print(image.shape[1])
```

```python
color = 0
# Row Gradient
for i in range(image.shape[1]):
    image[i,:] = i* (255/image.shape[1])

#Column Gradient
color = 0
for i in range(image.shape[1]):
    color_value = int(i * (255 / (image.shape[1] - 1)))
    image[:, i] = color_value
```

```python
print(image)
```

```python
cv2.imwrite('gradient.jpg', image)
```
