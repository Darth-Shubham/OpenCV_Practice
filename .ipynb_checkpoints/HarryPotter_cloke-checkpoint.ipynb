```python
import cv2
import numpy as np
```

```python
def setVal(val):
    print("")
```

```python
def trackbars():
    cv2.namedWindow("Trackbars")
    cv2.createTrackbar("Upper Hue", "Trackbars", 53, 180, setVal)
    cv2.createTrackbar("Upper Saturation", "Trackbars", 255, 255, setVal)
    cv2.createTrackbar("Upper Value", "Trackbars", 182, 255, setVal)
    cv2.createTrackbar("Lower Hue", "Trackbars", 12, 180, setVal)
    cv2.createTrackbar("Lower Saturation", "Trackbars", 17, 255, setVal)
    cv2.createTrackbar("Lower Value", "Trackbars", 21, 255, setVal)
```

```python
def get_frame_info(cap, scale_factor):
    success, img = cap.read()
    img = cv2.resize(img, None, fx = scale_factor, fy = scale_factor, interpolation = cv2.INTER_AREA)
    return img
```

```python
# Getting mask
def maskOperation(mask):
    # Creating Kernel
    kernel = np.ones((7,7), np.uint8)

    # Performing Opening operation(Erosion followed by Dilation)
    opening = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)
    
    # Performing Closing operation(Dilation followed by Erosion)
    closing = cv2.morphologyEx(opening, cv2.MORPH_CLOSE, kernel)
    
    return closing
```

```python
if __name__ == '__main__':
    cap = cv2.VideoCapture(0)
    scaleing_factor = 1.3
    trackbars()

    # Getting Initial Frame
    init_frame = get_frame_info(cap, scaleing_factor)
    
    while True:
        img = get_frame_info(cap, scaleing_factor)
        
        # Changing colour from BGR to HSV
        hsv_img = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

        # Getting Trackbar Upper values
        u_hue = cv2.getTrackbarPos("Upper Hue", "Trackbars")
        u_saturation = cv2.getTrackbarPos("Upper Saturation", "Trackbars")
        u_value = cv2.getTrackbarPos("Upper Value", "Trackbars")

        # Getting Trackbar Lower values
        l_hue = cv2.getTrackbarPos("Lower Hue", "Trackbars")
        l_saturation = cv2.getTrackbarPos("Lower Saturation", "Trackbars")
        l_value = cv2.getTrackbarPos("Lower Value", "Trackbars")

        # Upper value array
        upper_hsv = np.array([[u_hue, u_saturation, u_value]])
        
        # Lower value array       
        lower_hsv = np.array([[l_hue, l_saturation, l_value]])

        # Mask Range
        mask = cv2.inRange(hsv_img, lower_hsv, upper_hsv)
        
        mask = cv2.medianBlur(mask, 5)

        # Return value from maskOperation
        mask = maskOperation(mask)
        # Creating inverse of the mask
        mask_inv = 255-mask      

# -------------------------------------------------------------------------------------------------------------------------------------------------------
        # Splitting BGR channels from main Video
        b = img[:,:,0]
        g = img[:,:,1]
        r = img[:,:,2]
        # Performing AND operation on inverse of mask and BGR channels
        b = cv2.bitwise_and(mask_inv, b)
        g = cv2.bitwise_and(mask_inv, g)
        r = cv2.bitwise_and(mask_inv, r)
        # Final result will make the Cover area black and rest of the area is white
        # Merging Channels
        frame_inv = cv2.merge((b,g,r))
# -------------------------------------------------------------------------------------------------------------------------------------------------------

        b = init_frame[:,:,0]
        g = init_frame[:,:,1]
        r = init_frame[:,:,2]
        # Performing AND operation on inverse of mask and BGR channels
        b = cv2.bitwise_and(mask, b)
        g = cv2.bitwise_and(mask, g)
        r = cv2.bitwise_and(mask, r)
        # Final result will make the Cover area black and rest of the area is white
        # Merging Channels
        blank_area = cv2.merge((b,g,r))

# -------------------------------------------------------------------------------------------------------------------------------------------------------

        final_out = cv2.bitwise_or(frame_inv, blank_area)
        
        # cv2.imshow("original", img)
        cv2.imshow("mask", mask)
        # cv2.imshow('Frame', init_frame)
        cv2.imshow("image", final_out)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

cap.release()
cv2.destroyAllWindows()
```
