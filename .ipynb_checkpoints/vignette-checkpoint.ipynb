```python
import cv2
import numpy as np
```

```python
img = cv2.imread('face2.jpg')
img = cv2.resize(img, None, fx = .50 ,fy=.50)
height, width = img.shape[:2]
```

# Creating Vignette

```python
# Generating vignette mask using gausian kernel
kernel_x = cv2.getGaussianKernel(width, 180)
kernel_y = cv2.getGaussianKernel(height, 240)
kernel = kernel_y*kernel_x.T

# Normalizing Kernal
kernel = kernel/np.linalg.norm(kernel)

# Gererating image Mask
mask = kernel*255
final_img = np.copy(img)

for i in range(3):
    final_img[:,:,i] = final_img[:,:,i] * mask
```

```python
cv2.imshow('image', img)
cv2.imshow('masked image', final_img)

if cv2.waitKey(0) & 0xFF == ord('q'):
    cv2.destroyAllWindows()
```
