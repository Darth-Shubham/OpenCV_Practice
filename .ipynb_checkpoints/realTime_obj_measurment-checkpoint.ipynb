```python
import cv2
import numpy as np
%run Util_img_operations.ipynb
```

```python
webcam = False
cap = cv2.VideoCapture(0)
path = "a4_sheet.png"
path2 = "warp_A4.png"
scale = 2

wP = 210 * scale
hP = 297 * scale
```

```python
while True:
    if webcam: ret, img = cap.read()
    else: img = cv2.imread(path)
    # img = cv2.resize(img, (0,0), None, .25,.25)
    img, finalContours = grey_edgeDetect(img, cTh = [100,100], minArea=800, imgShow = False, ccd=True, filter=4, drawCont=False)

    if finalContours != 0:
        biggestCont = finalContours[0][1] # equal to Approx value
        # print(biggestCont)
        warpImg = getWarp(img, biggestCont, wP, hP) 
        
        img2 = cv2.imread(path2)
        img2, finalContours2 = grey_edgeDetect(img2, cTh = [100,100], minArea=800, imgShow = False, ccd=True, filter=0, drawCont=False)
        if finalContours2 != 0:
            for obj in finalContours2:
                cv2.polylines(img2, [obj[4]], True, (255,0,255), 2) # Using contour points
                
                newPoints = reorder(obj[1]) # Sending Approx value
                newObjWidth = round(findDist(newPoints[0][0]//scale, newPoints[1][0]//scale)/10, 1)
                newObjheight = round(findDist(newPoints[0][0]//scale, newPoints[2][0]//scale)/10, 1)

                cv2.arrowedLine(img2, (newPoints[0][0][0], newPoints[0][0][1]), 
                                (newPoints[1][0][0], newPoints[0][0][1]), (0,255,0), 2, 8, 0, 0.05)
                cv2.arrowedLine(img2, (newPoints[0][0][0], newPoints[0][0][1]), 
                                (newPoints[2][0][0], newPoints[2][0][1]), (0,255,0), 2, 8, 0, 0.05)
                x,y,w,h = obj[3]

                cv2.putText(img2, "{}cm".format(newObjWidth), (x+30, y-5) , cv2.FONT_HERSHEY_COMPLEX, 0.6, (0,255,0), 2)
                cv2.putText(img2, "{}cm".format(newObjheight), (x-60, y+30) , cv2.FONT_HERSHEY_COMPLEX, 0.6, (0,255,0), 2)
                
    cv2.imshow("original", img)
    cv2.imshow("warped Contour", img2)
    cv2.imshow("A4", warpImg)
    cv2.waitKey(1)
    if cv2.waitKey(1) & 0xff == ord('q'):
        break

cv2.destroyAllWindows()
```
