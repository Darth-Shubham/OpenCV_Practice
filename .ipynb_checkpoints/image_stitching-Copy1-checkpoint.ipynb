```python
import cv2
import numpy as np
import glob
import imutils
```

```python
image_path = glob.glob("images/*.jpg")
images = []
```

```python
for image in image_path:
    img = cv2.imread(image)
    img = cv2.resize(img, None, fx = .15, fy = .15)
    images.append(img)
    cv2.imshow("image",img)
    cv2.waitKey(0)
```

```python
# Creating stitcher object
imsticher = cv2.Stitcher_create()
```

```python
error, stichedImg = imsticher.stitch(images)
```

```python
if not error:
    cv2.imwrite('StichedOutput.jpg', stichedImg)
    cv2.imshow("stiched image", stichedImg)
    cv2.waitKey(0)
```
