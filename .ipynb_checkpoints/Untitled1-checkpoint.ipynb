```python
import cv2
import numpy as np
import glob
```

```python
path = "../Task/animals10/raw-img/cane/*"
```

```python
images = []
greyScale = []
path = glob.glob(path)
for img in path[0:10]:
    currImg = cv2.imread(img)
    greyimg = cv2.cvtColor(currImg.copy(), cv2.COLOR_BGR2GRAY)
    images.append(currImg)
    greyScale.append(greyimg)
```

```python
laplacian = cv2.Laplacian(greyScale[6], -1)
lap = cv2.morphologyEx(laplacian, cv2.MORPH_CLOSE, (7,7))
```

```python
thresh, mask = cv2.threshold(greyScale[6], 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
contour, _ = cv2.findContours(lap, mode = cv2.RETR_EXTERNAL, method = cv2.CHAIN_APPROX_SIMPLE)
```

```python
img_copy = cv2.drawContours(images[6].copy(), contour, -1, (0,255,0), thickness = 2, lineType = cv2.LINE_AA)
```

```python
max_contour = max(contour, key=cv2.contourArea)
x,y,w,h = cv2.boundingRect(max_contour)
```

```python
img_bound = cv2.rectangle(images[6].copy(), (x,y), (x+w, y+h), (0, 0 , 255), 2)
# img_bound = cv2.circle(img_bound, (x+(w//2),y+(h//2)), 5, (255, 0, 0), -1)
```

```python
cv2.imshow('Contour Detected',img_copy)
cv2.waitKey(0)
```

```python
import cv2
import numpy as np
import os
from tqdm import tqdm

# Paths
input_root = "../Task/animals10/raw-img/elefante/"
output_root = "../Task/animals10/masks/elefante/"

# Create output folders if not exist
os.makedirs(output_root, exist_ok=True)

# Parameters
border_thickness = 10
grabcut_iters = 5

for class_name in os.listdir(input_root):
    class_path = os.path.join(input_root, class_name)
    if not os.path.isdir(class_path):
        continue

    output_class_path = os.path.join(output_root, class_name)
    os.makedirs(output_class_path, exist_ok=True)

    for filename in tqdm(os.listdir(class_path), desc=f"Processing {class_name}"):
        if not filename.lower().endswith((".jpg", ".jpeg", ".png")):
            continue

        image_path = os.path.join(class_path, filename)
        image = cv2.imread(image_path)
        if image is None:
            print(f"Failed to load: {image_path}")
            continue

        h, w = image.shape[:2]
        mask = np.full((h, w), cv2.GC_PR_BGD, dtype=np.uint8)

        # Mark border as definite background
        mask[:border_thickness, :] = cv2.GC_BGD
        mask[-border_thickness:, :] = cv2.GC_BGD
        mask[:, :border_thickness] = cv2.GC_BGD
        mask[:, -border_thickness:] = cv2.GC_BGD

        # Mark center region as probable foreground
        center_h, center_w = h // 2, w // 2
        rect_size = min(h, w) // 4
        mask[
            center_h - rect_size : center_h + rect_size,
            center_w - rect_size : center_w + rect_size
        ] = cv2.GC_PR_FGD

        bgdModel = np.zeros((1, 65), np.float64)
        fgdModel = np.zeros((1, 65), np.float64)

        # Apply GrabCut
        cv2.grabCut(image, mask, None, bgdModel, fgdModel, grabcut_iters, cv2.GC_INIT_WITH_MASK)

        # Final mask: 1 for foreground, 0 for background
        result_mask = np.where((mask == cv2.GC_FGD) | (mask == cv2.GC_PR_FGD), 255, 0).astype(np.uint8)

        # Save the mask
        mask_filename = os.path.join(output_class_path, os.path.splitext(filename)[0] + "_mask.png")
        cv2.imwrite(mask_filename, result_mask)
```

```python
import cv2
import numpy as np
import os
from tqdm import tqdm

# Paths
input_root = "../Task/animals10/raw-img/"
output_root = "../Task/animals10/masks/"

# Create output root folder if not exist
os.makedirs(output_root, exist_ok=True)

# Parameters
border_thickness = 10
grabcut_iters = 5

# ===> Specify which class/folder to process
selected_class = "elefante"  # <-- change this to any class name you want

class_path = os.path.join(input_root, selected_class)
if not os.path.isdir(class_path):
    raise ValueError(f"Class folder '{selected_class}' not found in '{input_root}'")

output_class_path = os.path.join(output_root, selected_class)
os.makedirs(output_class_path, exist_ok=True)

for filename in tqdm(os.listdir(class_path), desc=f"Processing {selected_class}"):
    if not filename.lower().endswith((".jpg", ".jpeg", ".png")):
        continue

    image_path = os.path.join(class_path, filename)
    image = cv2.imread(image_path)
    if image is None:
        print(f"Failed to load: {image_path}")
        continue

    h, w = image.shape[:2]
    mask = np.full((h, w), cv2.GC_PR_BGD, dtype=np.uint8)

    # Mark border as definite background
    mask[:border_thickness, :] = cv2.GC_BGD
    mask[-border_thickness:, :] = cv2.GC_BGD
    mask[:, :border_thickness] = cv2.GC_BGD
    mask[:, -border_thickness:] = cv2.GC_BGD

    # Mark center region as probable foreground
    center_h, center_w = h // 2, w // 2
    rect_size = min(h, w) // 4
    mask[
        center_h - rect_size : center_h + rect_size,
        center_w - rect_size : center_w + rect_size
    ] = cv2.GC_PR_FGD

    bgdModel = np.zeros((1, 65), np.float64)
    fgdModel = np.zeros((1, 65), np.float64)

    # Apply GrabCut
    cv2.grabCut(image, mask, None, bgdModel, fgdModel, grabcut_iters, cv2.GC_INIT_WITH_MASK)

    # Final mask: 1 for foreground, 0 for background
    result_mask = np.where((mask == cv2.GC_FGD) | (mask == cv2.GC_PR_FGD), 255, 0).astype(np.uint8)

    # Save the mask
    mask_filename = os.path.join(output_class_path, os.path.splitext(filename)[0] + "_mask.png")
    cv2.imwrite(mask_filename, result_mask)
```
