```python
images = []
greyScale = []
path = glob.glob(path)
for img in path[0:10]:
    currImg = cv2.imread(img)
    greyimg = cv2.cvtColor(currImg.copy(), cv2.COLOR_BGR2GRAY)
    images.append(currImg)
    greyScale.append(greyimg)
```

```python
laplacian = cv2.Laplacian(greyScale[6], -1)
lap = cv2.morphologyEx(laplacian, cv2.MORPH_CLOSE, (7,7))
```

```python
thresh, mask = cv2.threshold(greyScale[6], 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
contour, _ = cv2.findContours(lap, mode = cv2.RETR_EXTERNAL, method = cv2.CHAIN_APPROX_SIMPLE)
```

```python
img_copy = cv2.drawContours(images[6].copy(), contour, -1, (0,255,0), thickness = 2, lineType = cv2.LINE_AA)
```

```python
max_contour = max(contour, key=cv2.contourArea)
x,y,w,h = cv2.boundingRect(max_contour)
```

```python
img_bound = cv2.rectangle(images[6].copy(), (x,y), (x+w, y+h), (0, 0 , 255), 2)
# img_bound = cv2.circle(img_bound, (x+(w//2),y+(h//2)), 5, (255, 0, 0), -1)
```

```python
import cv2
import numpy as np
import os
from tqdm import tqdm

# Paths
input_root = "../Task/animals10/raw-img/elefante/"
output_root = "../Task/animals10/masks/elefante/"

# Create output folders if not exist
os.makedirs(output_root, exist_ok=True)

# Parameters
border_thickness = 10
grabcut_iters = 5

for class_name in os.listdir(input_root):
    class_path = os.path.join(input_root, class_name)
    if not os.path.isdir(class_path):
        continue

    output_class_path = os.path.join(output_root, class_name)
    os.makedirs(output_class_path, exist_ok=True)

    for filename in tqdm(os.listdir(class_path), desc=f"Processing {class_name}"):
        if not filename.lower().endswith((".jpg", ".jpeg", ".png")):
            continue

        image_path = os.path.join(class_path, filename)
        image = cv2.imread(image_path)
        if image is None:
            print(f"Failed to load: {image_path}")
            continue

        h, w = image.shape[:2]
        mask = np.full((h, w), cv2.GC_PR_BGD, dtype=np.uint8)

        # Mark border as definite background
        mask[:border_thickness, :] = cv2.GC_BGD
        mask[-border_thickness:, :] = cv2.GC_BGD
        mask[:, :border_thickness] = cv2.GC_BGD
        mask[:, -border_thickness:] = cv2.GC_BGD

        # Mark center region as probable foreground
        center_h, center_w = h // 2, w // 2
        rect_size = min(h, w) // 4
        mask[
            center_h - rect_size : center_h + rect_size,
            center_w - rect_size : center_w + rect_size
        ] = cv2.GC_PR_FGD

        bgdModel = np.zeros((1, 65), np.float64)
        fgdModel = np.zeros((1, 65), np.float64)

        # Apply GrabCut
        cv2.grabCut(image, mask, None, bgdModel, fgdModel, grabcut_iters, cv2.GC_INIT_WITH_MASK)

        # Final mask: 1 for foreground, 0 for background
        result_mask = np.where((mask == cv2.GC_FGD) | (mask == cv2.GC_PR_FGD), 255, 0).astype(np.uint8)

        # Save the mask
        mask_filename = os.path.join(output_class_path, os.path.splitext(filename)[0] + "_mask.png")
        cv2.imwrite(mask_filename, result_mask)
```

```python
import os
import shutil
from tqdm import tqdm  # <-- for progress bar

# Translation dictionary
translate = {
    "cane": "dog", "cavallo": "horse", "elefante": "elephant",
    "farfalla": "butterfly", "gallina": "chicken", "gatto": "cat",
    "mucca": "cow", "pecora": "sheep", "scoiattolo": "squirrel", "ragno": "spider"
}

# Input folders
image_root = "../Task/animals10/raw-img/"
mask_root = "../Task/animals10/masks/"

# Output folders
renamed_image_root = "../Task/animals10/renamed"
renamed_mask_root = "../Task/animals10/renamed_masks"
os.makedirs(renamed_image_root, exist_ok=True)
os.makedirs(renamed_mask_root, exist_ok=True)

# Step 1: Manual selection
print("Available folders:")
for key in translate:
    print(f"- {key} ({translate[key]})")

selected = input("Enter folder names to rename (comma separated, in Italian): ").split(",")
selected = [name.strip() for name in selected]

# Step 2: Process selected folders
for italian_name in selected:
    if italian_name not in translate:
        print(f"⚠️ Skipping unknown folder: {italian_name}")
        continue

    english_name = translate[italian_name]
    img_folder = os.path.join(image_root, italian_name)
    mask_folder = os.path.join(mask_root, italian_name)

    if not os.path.isdir(img_folder):
        print(f"⚠️ Skipping missing image folder: {img_folder}")
        continue

    out_img_folder = os.path.join(renamed_image_root, english_name)
    out_mask_folder = os.path.join(renamed_mask_root, english_name)
    os.makedirs(out_img_folder, exist_ok=True)
    os.makedirs(out_mask_folder, exist_ok=True)

    # Get image files
    image_files = [f for f in os.listdir(img_folder) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
    counter = 1

    print(f"\n📂 Renaming images in '{italian_name}' ({english_name})")

    for filename in tqdm(image_files, desc=f"Processing {english_name}"):
        base_name = f"{english_name}_{counter:04d}"
        ext = os.path.splitext(filename)[1]

        img_path = os.path.join(img_folder, filename)
        mask_filename = os.path.splitext(filename)[0] + "_mask.png"
        mask_path = os.path.join(mask_folder, mask_filename)

        new_img_path = os.path.join(out_img_folder, base_name + ext)
        new_mask_path = os.path.join(out_mask_folder, base_name + "_mask.png")

        # Copy image
        shutil.copy(img_path, new_img_path)

        # Copy mask if it exists
        if os.path.exists(mask_path):
            shutil.copy(mask_path, new_mask_path)
        else:
            print(f"❌ Mask not found for: {filename}")

        counter += 1

print("\n✅ Done renaming selected folders.")
```
