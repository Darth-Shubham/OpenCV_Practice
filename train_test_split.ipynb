```python
import os
import shutil
from tqdm import tqdm
import random

# Set seed for reproducibility
random.seed(42)

# Paths
human_src = "../Task/human/images/"
animal_src = "../Task/animals10/renamed/"

# Destination structure
base_dest = "../Task/binary_dataset_split"
train_dest = os.path.join(base_dest, "train")
test_dest = os.path.join(base_dest, "test")

# Human & Animal class folders
train_human = os.path.join(train_dest, "Human")
test_human = os.path.join(test_dest, "Human")
train_animal = os.path.join(train_dest, "Animal")
test_animal = os.path.join(test_dest, "Animal")

# Create all destination directories
for path in [train_human, test_human, train_animal, test_animal]:
    os.makedirs(path, exist_ok=True)

# --- Copy Human Images ---
print("Processing human images...")
human_images = [f for f in os.listdir(human_src) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
random.shuffle(human_images)

for i, filename in enumerate(tqdm(human_images[:1000], desc="Humans")):
    src = os.path.join(human_src, filename)
    dest = os.path.join(train_human if i < 800 else test_human, filename)
    shutil.copy(src, dest)

# --- Copy Animal Images ---
print("\nProcessing animal images...")
animal_images = []

for species in tqdm(os.listdir(animal_src), desc="Collecting"):
    species_path = os.path.join(animal_src, species)
    if os.path.isdir(species_path):
        species_imgs = [f for f in os.listdir(species_path) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
        for img in species_imgs:
            animal_images.append((species, img))

random.shuffle(animal_images)

# Group by species, then limit to total 1000 animal images
species_dict = {}
for species, img in animal_images:
    species_dict.setdefault(species, []).append(img)

animal_images_limited = []
for species, imgs in species_dict.items():
    random.shuffle(imgs)
    animal_images_limited.extend([(species, img) for img in imgs[:1000 // len(species_dict)]])

# Trim to exactly 1000 images
animal_images_limited = animal_images_limited[:1000]
random.shuffle(animal_images_limited)

for i, (species, filename) in enumerate(tqdm(animal_images_limited, desc="Animals")):
    src = os.path.join(animal_src, species, filename)
    new_name = f"{species}_{filename}"
    dest = os.path.join(train_animal if i < 800 else test_animal, new_name)
    shutil.copy(src, dest)

print("\n✅ Dataset split into train (800) and test (200) for both classes!")
```

```python
import os
import shutil
from tqdm import tqdm
import random

# Set seed for reproducibility
random.seed(42)

# Paths
human_src = "../Task/human/images/"
animal_src = "../Task/animals10/renamed/"

# Destination structure
base_dest = "../Task/binary_dataset_split"
train_dest = os.path.join(base_dest, "train")
test_dest = os.path.join(base_dest, "test")

# Human & Animal class folders
train_human = os.path.join(train_dest, "Human")
test_human = os.path.join(test_dest, "Human")
train_animal = os.path.join(train_dest, "Animal")
test_animal = os.path.join(test_dest, "Animal")

# Create all destination directories
for path in [train_human, test_human, train_animal, test_animal]:
    os.makedirs(path, exist_ok=True)

# --- Copy Human Images ---
print("Processing human images...")
human_images = [f for f in os.listdir(human_src) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
random.shuffle(human_images)

selected_human_images = human_images[:600]

for i, filename in enumerate(tqdm(selected_human_images, desc="Humans")):
    src = os.path.join(human_src, filename)
    dest = os.path.join(train_human if i < 400 else test_human, filename)
    shutil.copy(src, dest)

# --- Copy Animal Images ---
print("\nProcessing animal images...")
animal_images_by_species = {}

# Collect images per species
for species in tqdm(os.listdir(animal_src), desc="Collecting species"):
    species_path = os.path.join(animal_src, species)
    if os.path.isdir(species_path):
        species_imgs = [f for f in os.listdir(species_path) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
        if len(species_imgs) >= 600:
            random.shuffle(species_imgs)
            animal_images_by_species[species] = species_imgs[:600]

# Copy images into Animal folders
for species, images in animal_images_by_species.items():
    print(f"\nProcessing species: {species} ({len(images)} images)")
    for i, filename in enumerate(tqdm(images, desc=f"{species}", leave=False)):
        src = os.path.join(animal_src, species, filename)
        new_name = f"{species}_{filename}"
        dest = os.path.join(train_animal if i < 400 else test_animal, new_name)
        shutil.copy(src, dest)

print("\n✅ Dataset created with:")
print(" - Human: 400 train, 200 test")
print(f" - Animal: {len(animal_images_by_species)} species, 400 train + 200 test each")
```
