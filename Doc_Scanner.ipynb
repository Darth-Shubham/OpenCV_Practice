```python
def greyScale(img):
    grey = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    grey_scale.append(grey)
    blurring(grey)
```

```python
def blurring(img):
    blurred = cv2.GaussianBlur(img, (5,5), 1)
    gaussian_blur.append(blurred)
    edgeDetection(blurred)
```

```python
def edgeDetection(img):
    cv2.namedWindow("TrackBar")
    cv2.createTrackbar("Lthresh", "TrackBar", 75, 255, empty)
    cv2.createTrackbar("Hthresh", "TrackBar", 135, 255, empty)
    
    while True:
        lthresh = cv2.getTrackbarPos("Lthresh", "TrackBar")
        hthresh = cv2.getTrackbarPos("Hthresh", "TrackBar")
                
        edge = cv2.Canny(img, lthresh, hthresh)
        cv2.imshow('Edge',edge)
        if cv2.waitKey(1) & 0xff == ord('q'):
            canny.append(edge)
            break
    cv2.destroyWindow('Edge')

# 36, 134 & 128(137), 255
# 23, 62
# 0, 255
# 0, 239 & 0, 255
# 17, 106, 17, 255
```

```python
def contour_area(img, img_copy):
    contour, hire = cv2.findContours(img, mode = cv2.RETR_EXTERNAL, method = cv2.CHAIN_APPROX_SIMPLE)
    high = np.array([])
    maxarea = 0
    for cnt in contour:
        area = cv2.contourArea(cnt)
        if area > 5000:
            
            img = cv2.drawContours(img_copy, cnt, -1, (0, 255, 0), 3)
            
            peri = cv2.arcLength(cnt, True)

            approx = cv2.approxPolyDP(cnt, 0.01*peri, True)
            
            if area > maxarea and len(approx) == 4:
                high = approx
                maxarea = area
                for point in approx:
                    x, y = point[0]
                    img = cv2.circle(img, (x, y), 5, (255, 0,0), -1)
                
    return high, img
        # peri = cv2.arcLength(cnt, True)
```

```python
def reorder(points):
    print(points.shape)
    
    points = points.reshape((4,2))
    newPoints = np.zeros((4,1,2), np.int32)
    
    add = points.sum(1)
    newPoints[0] = points[np.argmin(add)]
    newPoints[3] = points[np.argmax(add)]

    diff = np.diff(points, axis = 1)
    newPoints[1] = points[np.argmin(diff)]
    newPoints[2] = points[np.argmax(diff)]

    return newPoints
    
```

```python
def getWarp(img, area):
    points = reorder(area)
    source_points = np.float32(points)
    dest_points = src_points = np.float32([[0,0], [width,0], [0,height], [width,height]])
    matrix = cv2.getPerspectiveTransform(source_points, dest_points)
    img_out = cv2.warpPerspective(img, matrix, (width, height))

    cropped_img = img_out[20:img_out.shape[0]-10, 20:img_out.shape[1]-10]
    return cropped_img
```

```python
for doc in docs_path:
    document = cv2.imread(doc)
    
    # Get original image dimensions
    height, width = document.shape[:2]
    # Calculate aspect ratios
    aspect_ratio = width / height
    
    if width > screen_width or height > screen_height:
    # Calculate new dimensions based on screen size and aspect ratio
        if width / screen_width > height / screen_height:
            new_width = screen_width
            new_height = int(new_width / aspect_ratio) - 100
        else:
            new_height = screen_height - 100
            new_width = int(new_height * aspect_ratio)
    
        # Rzesize the image
        resized_document = cv2.resize(document, (new_width, new_height))
    else:
        # If image is smaller than screen size, keep it as is
        resized_document = document
    print(resized_document.shape)
    docs.append(resized_document)
    cv2.imshow('document', resized_document)
    if cv2.waitKey(0) & 0xff==ord('q'):
        cv2.destroyAllWindows()

```
