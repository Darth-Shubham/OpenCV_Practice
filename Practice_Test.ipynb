## Non-Maximum Suppression

```python
# cv2.rectangle(img, ((width//2), (height//2)-100), ((width//2)+100, (height//2)), (255,0,0), 2)
# cv2.rectangle(img, ((width//2)-45, (height//2)-130), ((width//2)+35, (height//2)-35), (0,255,0), 2)
# cv2.rectangle(img, ((width//2)-15, (height//2)-120), ((width//2)+80, (height//2)-15), (0,0,255), 2)
# # cv2.rectangle(img, ((width//2), (height//2)-100), ((width//2)+100, (height//2)), (255,255,0), 2)


# Assuming width and height are defined
bbox1 = [width // 2, (height // 2) - 100, 100, 100]                        # Blue
bbox2 = [(width // 2) - 45, (height // 2) - 130, 80, 95]                  # Green
bbox3 = [(width // 2) - 15, (height // 2) - 120, 95, 105]                 # Red

bboxes = [bbox1, bbox2, bbox3]
scores = [0.7, 0.85, 0.9] 
```

```python
score_thresh = 0.5
nms_thresh = 0.4

# Run NMS
indices = cv2.dnn.NMSBoxes(bboxes, scores, score_thresh, nms_thresh)
# Flatten and select boxes
indices
selected_boxes = [bboxes[i] for i in indices]
print("Boxes kept after NMS:", selected_boxes)
```

```python
for box in selected_boxes:
    x, y, w, h = box
    top_left = (x, y)
    bottom_right = (x + w, y + h)
    cv2.rectangle(img, top_left, bottom_right, (255, 255, 0), 2)
```
