# Image Filter

```python
#importing libs
import cv2
import numpy as np
```

```python
#Reading the omage
img = cv2.imread('noisy.jpg')
height, width = img.shape[0:2]
```

# Blur Flilters (Low Pass)

## Kernel Bluring/Low Fliter(Bluring the image)

```python
# Creating a lowpass kernel(Filter)
filter = np.ones((5,5), dtype = np.float32)
# Normalizing Filter
filter = filter/filter.sum()
```

```python
#Applying the Filter
output = cv2.filter2D(img, -1, filter)
```

## Box Filter and Blur Function

```python
# Using Blur function
blur_func = cv2.blur(img, (5,5))
```

```python
# Using Box function
box_func = cv2.boxFilter(img, -1, (5,5), normalize = False)
```

## Gaussian Blur

```python
gaus_func = cv2.GaussianBlur(img, (5,5), 0)
```

## Median Blur

```python
median_func = cv2.medianBlur(img, 5)
```

## Bilateral Filtering

```python
bilateral_func = cv2.bilateralFilter(img, 5, 6, 6)
# src = original_image, d= radius of kernel, 
# sigmaColor = standard deviation of color(makes similar intensity pixels as one), 
# sigmaSpace = same as sigmaColor
```

## Motion blur

```python
size = 15
# Creating kernel for horizontal motion blur
motion_filter = np.zeros((size, size))
motion_filter[int((size-1)/2),:] = np.ones(size)
# Normalizing Filter
motion_filter = motion_filter/size
```

```python
# applying motion blur filter
motion_output = cv2.filter2D(img, -1, motion_filter)
```

## Showing Images

```python
cv2.imshow("Original Image", img)
cv2.imshow("Kernel Filtering", output)
cv2.imshow("Blur() Filtering", blur_func)
cv2.imshow("Box() Filtering", box_func)
cv2.imshow("Gaussian Blur", gaus_func)
cv2.imshow("Median Blur", median_func)
cv2.imshow("Bilateral Blur", bilateral_func)

if cv2.waitKey(0) & 0xFF == ord('q'):
    cv2.destroyAllWindows()
```

# Image Sharpening (High Pass)

```python
# Taking Blur of and image using Gaussian blur
blur = cv2.GaussianBlur(img, (7, 7), 3)
```

```python
# Applying Sharpening
sharpen_1 = cv2.addWeighted(img, 2, blur, -1, 0)
sharpen_2 = cv2.addWeighted(img, 3.5, blur, -2.5, 0)
sharpen_3 = cv2.addWeighted(img, 7.5, blur, -6.5, 0)
```

```python
cv2.imshow("Original Image", img)
cv2.imshow("sharpen_1 Filtering", sharpen_1)
cv2.imshow("sharpen_2 Filtering", sharpen_2)
cv2.imshow("sharpen_3 Filtering", sharpen_3)


if cv2.waitKey(0) & 0xFF == ord('q'):
    cv2.destroyAllWindows()
```
