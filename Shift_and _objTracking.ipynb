```python
import cv2
import numpy as np
```

```python
cap = cv2.VideoCapture('../Yolo_projects/Videos/car_count.mp4')
_, frame = cap.read()
x, y, w, h = 750, 348, 125, 106
track_window = (x, y, w, h)
```

```python
roi = frame[y: y+h, x: x+w]
hsv_roi = cv2.cvtColor(roi, cv2.COLOR_BGR2HSV)
mask = cv2.inRange(hsv_roi, np.array((0, 60, 32)), np.array((180, 255, 255)))
roi_hist = cv2.calcHist([hsv_roi], [0], mask, [108], [0,180])
cv2.normalize(roi_hist, roi_hist, 0, 255, cv2.NORM_MINMAX)
```

```python
term_criteria = (cv2.TermCriteria_EPS | cv2.TERM_CRITERIA_COUNT, 10, 1)
```

```python
while True:
    ret, frame = cap.read()
    if ret == False:
        cv2.waitKey(0)
        break
    dst = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)
    dst = cv2.calcBackProject([dst], [0], roi_hist, [0,180], 1)
    _, track_window = cv2.meanShift(dst, track_window, term_criteria)

    x,y,w,h = track_window

    final_img = cv2.rectangle(frame, (750, 348), (750 + 125, 348 + 106), (0,255,255), 2)

    cv2.imshow('result', final_img)

    if cv2.waitKey(0) & 0xff==ord('q'):
        break

cv2.destroyAllWindows()
```
