```python
import cv2
import numpy as np
```

```python
vid = cv2.VideoCapture('../Yolo_projects/Videos/ppe1.mp4')
fgbgMOG = cv2.bgsegm.createBackgroundSubtractorMOG()
fgbg = cv2.createBackgroundSubtractorKNN(history = 1,dist2Threshold = 1000)
```

```python
while True:
    ret, frame = vid.read()
    
    if frame is None:
        cv2.waitKey(0)
        break
    fgmaskMOG = fgbgMOG.apply(frame)
    fgmask = fgbg.apply(frame)
    
    cv2.imshow('Vid', frame)
    cv2.imshow('fg mask', fgmask)
    cv2.imshow('fg mask MOG', fgmaskMOG)
    
    if cv2.waitKey(2) & 0xff == ord('q'):
        break
    
cv2.destroyAllWindows()
```
