```python
import cv2
import numpy as np
```

```python
kernel = np.ones((3,3))
```

```python
def frame_diff(prev_frame, current_frame, next_frame):
    # Finding absolute difference between current_frame & prev_frame
    diff_frame1 = cv2.absdiff(current_frame, prev_frame)
    # Finding absolute difference between current_frame & next_frame
    diff_frame2 = cv2.absdiff(next_frame, current_frame)
    # Returning the result of bitwise AND opperation
    return cv2.bitwise_and(diff_frame1, diff_frame2)
```

```python
def get_frame(cap, scaling_fact):
    success, img = cap.read()
    # Resizing the incoming frame
    img = cv2.resize(img, None, fx=scaling_fact, fy = scaling_fact, interpolation=cv2.INTER_AREA)
    return cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
if __name__ == '__main__':
    # cap = cv2.VideoCapture('../Yolo_projects/Videos/escelator3.mp4')
    cap = cv2.VideoCapture('rev.mp4')

    scaling_fact = .8
    
    # Getting the first 3 Frames
    prev_frame = get_frame(cap, scaling_fact)
    current_frame = get_frame(cap, scaling_fact)
    next_frame = get_frame(cap, scaling_fact)

    while True:
        # Sending first 3 frames to frame_diff function
        frame = frame_diff(prev_frame, current_frame, next_frame)
        # frame = cv2.morphologyEx(frame, cv2.MORPH_CLOSE, np.ones((5,5)), iterations = 1)
        # Getting the mask of moving objects
        _, frame_th = cv2.threshold(frame, 0, 255, cv2.THRESH_TRIANGLE)
        cv2.imshow("object Movement", frame)
        cv2.imshow("Obj", frame_th)

        # changing the frames to next frame
        prev_frame = current_frame
        current_frame = next_frame
        try:
            next_frame = get_frame(cap, scaling_fact)
        except Exception as e:
            print("End of Video")
            if cv2.waitKey(0) & 0xFF == ord('q'):
                break
        
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break
    cv2.destroyAllWindows()
```
