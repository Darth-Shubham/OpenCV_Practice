```python
import numpy as np
import cv2
```

```python
# Rectangle Canvas
rec_canvas = np.full((250, 500), 0, np.uint8)
# Square canvas
sqr_canvas = np.full((500,500), 0, np.uint8)
```

```python
rec_height, rec_width = rec_canvas.shape[:2]
sqr_height, sqr_width = sqr_canvas.shape[:2]
```

# Gradient Study

```python
# Horzontal Color Gradient
for y in range(rec_height):
    rec_canvas[y,:] = int((y / (rec_height - 1)) * 255)
```

```python
# Vertical Color Gradient
for x in range(rec_width):
    rec_canvas[:,x] = int((x / (rec_width - 1)) * 255)
```

```python
# Reverse Colour(White - Black) Horizontal Gradient
for y in range(rec_height):
    rec_canvas[y,:] = int(255 - 255*(y/(rec_height-1)))
```

```python
# Diagonal Gradient Rectangle
for y in range(rec_height):
    for x in range(rec_width):
        rec_canvas[y,x] = int(((x+y)/(rec_height+rec_width - 2)) * 255)
```

```python
# Diagonal Gradient Square
for y in range(sqr_height):
    for x in range(sqr_width):
        sqr_canvas[y,x] = int(((x + y) / (sqr_height+sqr_width - 2)) * 255)
```

```python
# Mirror Diagonal Gradient Square
for y in range(sqr_height):
    for x in range(sqr_width):
        sqr_canvas[y,x] = 255 - abs(((x+y)/(sqr_height+sqr_width-2)) - .5) * 510
```

# Functions

```python
def straight_gradients(canvas, direction = 'verti', gradient = 'b_w'):
    height, width = canvas.shape[:2]
    if direction == 'hori':
        if gradient == 'b_w':
            for y in range(height):
                canvas[y, :] = int((y / (height - 1)) * 255)
        else:
            for y in range(height):
                canvas[y, :] = 255 - int((y / (height - 1)) * 255)
    
    elif direction == 'verti':
        if gradient == 'b_w':
            for x in range(width):
                canvas[:, x] = int((x / (width - 1)) * 255)
        else:
            for x in range(width):
                canvas[:, x] = 255 - int((x / (width - 1)) * 255)
    
    else:
        print('direction should be either "verti" or "hori"')

    return canvas
```

```python
def diagonal_gradients(canvas, mirror = 'no'):
    height, width = canvas.shape[:2]
    if mirror == 'no':
        for y in range(height):
            for x in range(width):
                canvas[y,x] = int(((y + x)/(height + width - 2)) * 255)
                
    elif mirror == 'yes':
        for y in range(height):
            for x in range(width):
                canvas[y,x] = 255 - abs(((y+x)/(height + width - 2))-.5) * 510
                
    else:
        print('mirror should be either yes or no')

    return canvas
```

```python
channle1 = straight_gradients(rec_canvas.copy(), direction='verti', gradient='1')
channle2 = straight_gradients(rec_canvas.copy(), direction='verti')
channle3 = diagonal_gradients(rec_canvas.copy(), 'yes')
```

```python
final_img_rec = cv2.merge((channle1, channle2, channle3))
cv2.imshow('canvas', final_img_rec)
cv2.waitKey(0)
```
