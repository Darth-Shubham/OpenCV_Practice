# Reversing a Video

```python
#initializing video reader
cap = cv2.VideoCapture('../Yolo_projects/Videos/car_count.mp4')

#getting video information
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
fps = int(cap.get(cv2.CAP_PROP_FPS))
frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
res = kheight)

# initializing video writter
fourcc = cv2.VideoWriter_fourcc(*'MP4V') # initializing the codec
out = cv2.VideoWriter('rev.mp4', fourcc, fps, res)
```

```python
print(height, width, fps, frames)
```

```python
last_frame = frames - 1
rotation_increment = 360/frames
frame_count = 0
```

```python
while last_frame>=0:
    #setting the video start to last frame
    cap.set(cv2.CAP_PROP_POS_FRAMES, last_frame)
    #readign the video
    success, img = cap.read()
    if success:
        # Video Rotation Logic
        rotation_angle = (frames - last_frame) * rotation_increment
        # Rotation Matrix
        rotation_matrix = cv2.getRotationMatrix2D((width/2, height/2), rotation_angle, 1)
        # Adding Rotation to the video
        rotated = cv2.warpAffine(img, rotation_matrix, (width, height))
    
        # To show the video in window
        # cv2.imshow("image", img)
        # if cv2.waitKey(1) & 0xFF == ord('q'):
        #     break
    
        # Writting the reversed video
        out.write(rotated)

        frame_count += 1

        # Print progress every 100 frames
        if frame_count % 100 == 0:
            print(f'Frames written: {frame_count}/{frames} ({(frame_count / frames) * 100:.2f}%)')
    
        # moving the last_frame counter by 1
        last_frame = last_frame - 1
    

# Release the video capture and close any OpenCV windows
cap.release()
out.release()
cv2.destroyAllWindows()
```

# Drawing Coloured Lines, Shapes and Texts

```python
canvas = np.zeros((500,500, 3)) # 3 for RGB channel
```

## Creating Lines(aliasing - LN_4, LN_8 & anti aliasing - LN_AA)

```python
# Line 4 & 8 - bresenham
cv2.line(canvas, (25,25), (475, 25), (255,255,255), 3, cv2.LINE_4)
# Line AA - Gaussian Filtering
cv2.line(canvas, (25,325), (475, 325), (255,255,255), 3, cv2.LINE_AA)

# cv2.imshow('image', canvas)
# cv2.waitKey(0)
```

## Creating Rectangle

```python
cv2.rectangle(canvas, (120,40), (195, 310), (120,150,0), 1, cv2.LINE_4) #thickness = -1 for filled rectangle
# cv2.imshow('image', canvas)
# cv2.waitKey(0)
```

## Creating Arrow Line

```python
cv2.arrowedLine(canvas, (215,175), (235, 175), (255,255,255), 2, cv2.LINE_AA, tipLength = .2)
# cv2.imshow('image', canvas)
# cv2.waitKey(0)
```

## Creating Circles

```python
cv2.circle(canvas, (330,175), 75, (120,0,120), -1, cv2.LINE_8) #thickness = -1 for filled rectangle
canvas[175, 237] = 255
cv2.imshow('image', canvas)
cv2.waitKey(0)
```

# Drawing PolyLines

```python
canvas = np.zeros((300,300,3))
```

```python
# Point to plot on canvas
pts = np.array([[10,5], [30, 25], [50, 5], [30, 50]], np.int32)

#reshaping points to shape (number_vertex, 1, 2)
pts = pts.reshape(-1,1,2)

# Drawing Polygon
cv2.polylines(canvas, [pts], True, (255,255,255)) # Here Boolean(True, False) determines if figure is closed(True) or not(False)

cv2.imshow('Image', canvas)
cv2.waitKey(0)
```
