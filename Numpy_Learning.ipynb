```python
import numpy as np
```

# Matrix Dimention and Orientation

## 1D Arrays

```python
# Row Vector
arr = np.array([1,2,3,4,5])
print(arr)
# Column Vector (has to be a matrix with atleast 2 dimentions)
arr_v = np.array([[1],[2],[3],[4],[5]])
print(arr_v)

# finding the dimention of the vector is
print(f"Diemention of arr  - {arr.ndim}")
print(f"Diemention of arr_v - {arr_v.ndim}")
```

# N Dimention Arrays

```python
arr_3D = np.array([[[1,2,3,4,5,6], [6,7,8,9,0,12]]])
print(arr_3D)
print(arr_3D.ndim)
```

# # Fill array of N dimentions with 0

```python
# Creating zero Array
zero_arr = np.zeros((4,5)) # by default created the array of type float
print(zero_arr)

# for creating and integer zero array
zero_arr_i = np.zeros((4,5), dtype = 'int8')
print(zero_arr_i)
```

# Fill array of N dimentions with 1

```python
# Creating ones Array
one_arr = np.ones((4,5)) # by default created the array of type float
print(one_arr)

# for creating and integer zero array
one_arr_i = np.ones((4,5), dtype = 'int8')
print(one_arr_i)
```

# Fill array of N dimentions with a custom Value

```python
cus_arr = np.full((2,3), 5) # automatically takes the data type of the value
print(cus_arr)
```

# Creating identity matrix

```python
eye_arr = np.eye(5, dtype = 'int8')
print(eye_arr)
```

# Auto creating Array

```python
a_arr = np.arange(0,14,3, dtype = 'int8')
print(a_arr)
```

# Creating array with N values with N distance

```python
l_arr = np.linspace(0,5,11)
print(l_arr)
```
